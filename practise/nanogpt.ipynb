```python
def encode(st):
    return [stoi(s) for s in st]

def decode (it):
    return ''.join(itos[i] for i in it)
```

```python
stoi = {ch:i for i,ch in enumerate(chars)}
itos = {i:ch for i,ch in enumerate(chars)}
encode = lambda st: [stoi[s] for s in st]
decode = lambda it: [''.join(itos[i] for i in it)]
encode("Hello"),"==>",decode(encode("Hello"))
```
